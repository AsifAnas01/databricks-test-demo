```python
# Load tables with backtick-quoted catalog name
departments_df = spark.table("`my-test`.default.departments")
employees_df = spark.table("`my-test`.default.employees")


# Join on DEPARTMENT_ID
joined_df = employees_df.join(
    departments_df,
    employees_df.EMPLOYEE_ID == departments_df.DEPARTMENT_ID,
    "inner"
)

# Transformations:
# 1. Select relevant columns and rename DEPARTMENT_NAME
# 2. Add a column for employee full name
# 3. Filter employees with salary > 50000
transformed_df = (
    joined_df
    .select(
        employees_df.EMPLOYEE_ID,
        employees_df.FIRST_NAME,
        employees_df.LAST_NAME,
        employees_df.SALARY,
        departments_df.DEPARTMENT_NAME.alias("DEPT_NAME")
    )
    .filter(employees_df.SALARY > 50000)
)

display(transformed_df)
```

```python
%python
from pyspark.sql.functions import when

# Read the departments table with backtick-quoted catalog name
df = spark.read.table("`my-test`.default.departments")

# Transformation 1: Rename DEPARTMENT_NAME to DEPT_NAME
df1 = df.withColumnRenamed("DEPARTMENT_NAME", "DEPT_NAME")

# Transformation 2: Filter departments with MANAGER_ID not null
df2 = df1.filter(df1["MANAGER_ID"].isNotNull())

# Transformation 3: Add a new column 'LOCATION_TYPE' based on LOCATION_ID
df3 = df2.withColumn(
    "LOCATION_TYPE",
    when(
        df2["LOCATION_ID"] > 200,
        "Remote"
    ).otherwise("Onsite")
)

# Transformation 4: Select specific columns and reorder
final_df = df3.select(
    "DEPARTMENT_ID",
    "DEPT_NAME",
    "MANAGER_ID",
    "LOCATION_TYPE"
)

display(final_df)
```

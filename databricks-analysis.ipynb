```python
from pyspark.sql.functions import concat_ws

df = spark.table("`my-test`.default.employees")

df_filtered = df.filter(df.SALARY > 50000)

df_fullname = df_filtered.withColumn(
    "FULL_NAME",
    concat_ws(
        " ",
        df_filtered.FIRST_NAME,
        df_filtered.LAST_NAME
    )
)

df_annual = df_fullname.withColumn(
    "ANNUAL_SALARY",
    df_fullname.SALARY * 12
)

df_final = df_annual.select(
    "EMPLOYEE_ID",
    "FULL_NAME",
    "EMAIL",
    "ANNUAL_SALARY"
).orderBy(
    df_annual.ANNUAL_SALARY.desc()
)

display(df_final)
```
